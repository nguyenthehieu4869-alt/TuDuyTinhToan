```python
#W6A10
a = list(map(int, input().split()))
k = int(input())
b = []
for i in range(len(a)):
    for y in range(i+1, len(a)):
        if a[i] + a[y] == k:
            c = (a[i], a[y])
            if c not in b:
                b.append(c)
print(b)
```

```python
#W6A14
def compare(x, y):
    c = set()
    for i in x:
        if i in y:
            c.add(i)
    return list(c)        
a = list(map(int, input().split()))
b = list(map(int, input().split()))
if len(a) >= len(b):
    print(compare(b, a))
else:
    print(compare(a, b))
```

```python
#W6A16
arr = []
n, m = map(int, input("Nhập vào số hàng (n), số cột (m): ").split())
for i in range(n):
    one_row = []
    for j in range(m):
        one_row.append(int(input(f"Nhập vào hàng {i+1}, cột {j+1}: ")))
    arr.append(one_row)
for q in arr:
    for k in q:
        print("{:>4}".format(k), end = "")
    print()
```

```python
#W6A17
row = column = int(input())
#Main_diag
for i in range(row):
    for j in range(column):
        if i == j:
            element = int(input(f"Nhập vào phần tử ở hàng {i+1}, cột {j+1}: "))
            print(element, end = "")
        else:
            print("   ", end = "")
    print()
#Side_diag
row = column = int(input())
for q in range(1, row + 1):
    for k in range(1, column + 1 ):
        if q + k == row + 1:
            element = int(input(f"Nhập vào phần tử ở hàng {q}, cột {k}: "))
            print(element, end = "")
        else:
            print("   ", end = "")
    print()    


        
```

```python
#W6A18
n, m, k = map(int, input().split())
if k > n and k > m:
    print(0)
elif k > n and k <= m:
    cnt = []
    for i in range(n):
        for j in range(m):
            if j + 1 == k:
                cnt.append(int(input(f"Nhập vào phần tử ở hàng {i + 1}, cột {k}: " )))
    print(f"Tổng các phần tử trên cột {k}: {sum(cnt)}")
elif k > m and k <= n:
    cnt = []
    for i in range(n):
        for j in range(m):
            if i + 1 == k:
                cnt.append(int(input(f"Nhập vào phần tử ở hàng {k}, cột {j + 1}: " )))
    print(f"Tổng các phần tử trên hàng {k}: {sum(cnt)}")
else:
    cnt = []
    for i in range(n):
        for j in range(m):
            if i + 1 == k:
                cnt.append(int(input(f"Nhập vào phần tử ở hàng {k}, cột {j + 1}: " )))
    print(f"Tổng các phần tử trên hàng {k}: {sum(cnt)}")
    cnt.clear()
    for q in range(n):
        for p in range(m):
            if p + 1 == k:
                cnt.append(int(input(f"Nhập vào phần tử ở hàng {q + 1}, cột {k}: " )))
    print(f"Tổng các phần tử cột {k}: {sum(cnt)}")
```
