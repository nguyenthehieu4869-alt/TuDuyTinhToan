```python
#W8A1
class Cylinder:
    def __init__(self, radius, height, pi = 3.14):
        self.radius = radius
        self.height = height
        self.pi = 3.14
    def surface_area(self):
        return 2 * self.pi * (self.radius ** 2) + 2 * self.pi * self.radius * self.height
    def volume(self):
        return self.pi * (self.radius ** 2) * self.height
r, h = map(int, input().split())
cylinder = Cylinder(r, h)
surface_area = cylinder.surface_area()
volume = cylinder.volume()
print(f"{surface_area:.2f} {volume:.2f}")
```

```python
#W8A2
class Date:
    def __init__(self, day, month, year):
        self.d = day
        self.m = month
        self.y = year
    def check_leap_year(self):
        return self.y % 400 == 0 or self.y % 4 == 0 and self.y % 100 != 0
    def check_day_month(self):
        if self.m < 1 or self.m > 12:
            return 0
        elif self.m == 2:
            return 29 if self.check_leap_year() else 28
        elif self.m in [4, 6, 9, 11]:
            return 30
        return 31
    def next_day(self):
        if self.d < 1 or self.d > self.check_day_month():
            return "INVALID"
        else:
            self.d += 1
            if self.d > self.check_day_month():
                self.d = 1
                self.m += 1
                if self.m > 12:
                    self.m = 1
                    self.y += 1
        return self
    def __str__(self):
        return f"{self.d:02d}/{self.m:02d}/{self.y}"
d, m, y = map(int, input().split())
date = Date(d, m, y)
print(date.next_day())
```

```python
#W8A3
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def check_position(self):
        if self.x == 0 and self.y == 0:
            return "M la goc toa do"
        elif self.x == 0 and self.y != 0:
            return "M thuoc truc tung"
        elif self.x != 0 and self.y == 0:
            return "M thuoc truc hoanh"
        else:
            return "M khong thuoc truc toa do"
    def distance(self):
        return (self.x ** 2 + self.y ** 2) ** 0.5
x, y = map(int, input().split())
point = Point(x, y)
vi_tri = point.check_position()
k_cach = point.distance()
print(vi_tri, f'{k_cach:.2f}', sep='\n')
```

```python
#W8A5
class ShoppingCart:
    def __init__(self):
        self.items = []
    def add_item(self, name, price):
        self.items.append({"name": name, "price": price})
    def remove_item(self, name):
        for i, item in enumerate(self.items):
            if item["name"] == name:
                self.items.pop(i)
                return True
        return False
    def check_empty_item(self):
        return len(self.items) == 0
    def total_cost_items(self):
        total = 0
        for i in self.items:
            total += i["price"]
        return total
    def show_all_items(self):
        if self.check_empty_item():
            print("Gio hang trong")
        else:
            for i in self.items:
                print(f"{i["name"]}", ":", f"{i["price"]}")
    def clear_all_items(self):
        self.items.clear()
cart = ShoppingCart()
print("------ MENU ------")
print("2. Them san pham")
print("3. Xoa san pham theo ten")
print("4. Kiem tra gio hang rong")
print("5. Tinh tong tien")
print("6. Hien thi tat ca san pham")
print("7. Xoa toan bo gio hang")
while True:
    c = input()
    if c == "2":
        name = input()
        price = float(input())
        cart.add_item(name, price)
    elif c == "3":
        name = input()
        if cart.remove_item(name):
            print("Da xoa san pham")
        else:
            print("Khong tim thay san pham")
    elif c == "4":
        print("Gio hang rong" if cart.check_empty_item() else "Gio hang co san pham")
    elif c == "5":
        print(f"Tong tien: {cart.total_cost_items():.2f}")
    elif c == "6":
        cart.show_all_items()
    elif c == "7":
        cart.clear_all_items()
    else:
        print("Lua chon khong hop le")
    d = input("Ban co muon thoat?")
    if d.lower() == "yes":
        break
```

```python
#W8A6
class Rectangle:
    def __init__(self, w, h):
        self.w = w
        self.h = h
    def scale(self, k):
        self.w *= k
        self.h *= k
    def area(self):
        return self.w * self.h
    def perimeter(self):
        return 2 * (self.w + self.h)
w, h, k = map(float, input().split())
rec = Rectangle(w, h)
rec.scale(k)
area = rec.area()
perimeter = rec.perimeter()
print(f'{area:.2f}, {perimeter:.2f}')
```

```python
#W8A7
import math 
class Fraction:
    def __init__(self, num, den):
        self.n = num
        self.d = den
        if self.d == 0:
            raise ValueError("Khong the chia cho 0")
        if self.d < 0:
            self.d *= -1
            self.n *= -1
        self.reduce()
    def reduce(self):
        g = math.gcd(abs(self.n), abs(self.d))
        self.n //= g
        self.d //= g
    def add(self, other):
        num = self.n * other.d + self.d * other.n
        den = self.d * other.d
        return Fraction(num, den)
    def subtract(self, other):
        num = self.n * other.d - self.d * other.n
        den = self.d * other.d
        return Fraction(num, den)
    def multiply(self, other):
        num = self.n * other.n
        den = self.d * other.d
        return Fraction(num, den)
    def divide(self, other):
        if other.n == 0:
            raise ValueError("Khong the chia cho 0")
        else:
            num = self.n * other.d
            den = self.d * other.n
            return Fraction(num, den)
    def __str__(self):
        return f"{self.n}/{self.d}"
a, b, op, c, d = input().split()
a, b, c, d = int(a), int(b), int(c), int(d)
f1 = Fraction(a, b)
f2 = Fraction(c, d)
if op == "+":
    print(f1.add(f2))
elif op == "-":
    print(f1.subtract(f2))
elif op == "*":
    print(f1.multiply(f2))
elif op == "/":
    print(f1.divide(f2))
else:
    print("Phep tinh khong hop le")
```

```python
#W8A8
class BankAccount:
    def __init__(self, owner, balance):
        self.o = owner
        self.b = balance
    def deposit(self, x):
        self.b += x
    def withdraw(self, x):
        if self.b >= x:
            self.b -= x
    def __str__(self):
        return f"{self.b:.2f}"
o, b = input().split()
b= float(b)
n = int(input())
bankAcc = BankAccount(o, b)
for _ in range(n):
    a, x = input().split()
    x = float(x)
    if a.lower() == "deposit":
        bankAcc.deposit(x)
    elif a.lower() == "withdraw":
        bankAcc.withdraw(x)
print(bankAcc)        
```

```python
#W8A9
class Student:
    def __init__(self, name):
        self.n = name
        self.total = 0
        self.count = 0
    def add_score(self, score):
        self.total += score
        self.count += 1
    def avg_score(self):
        return self.total / self.count
    def rank(self):
        a = self.avg_score()
        if a >= 8:
            return "Excellent"
        elif 6.5 <= a < 8:
            return "Good"
        elif 5 <= a < 6.5:
            return "Average"
        else:
            return "Poor"
a = input()
stu = Student(a)
n = int(input())
for _ in range(n):
    s, b = input().split()
    b = float(b)
    stu.add_score(b)
print(stu.n, f"{stu.avg_score():.2f}", stu.rank())
```

```python
#W8A11
import math
class Point:
    def __init__(self, x, y, z):
        self.x = x
        self.y = y
        self.z = z
    def vectors(self, other):
        return Vector(
            other.x - self.x,
            other.y - self.y,
            other.z - self.z
        )
class Vector:
    def __init__(self, x, y, z):
        self.x = x
        self.y = y
        self.z = z
    def tich_co_huong(self, other):
        return Vector(
            self.y * other.z - self.z * other.y,
            self.z * other.x - self.x * other.z,
            self.x * other.y - self.y * other.x
        )
    def tich_vo_huong(self, other):
        return self.x * other.x + self.y * other.y + self.z * other.z
    def tich_do_dai(self):
        return math.sqrt(self.x ** 2 + self.y ** 2 + self.z ** 2)
class Plane:
    def __init__(self, p1, p2, p3):
        v1 = p1.vectors(p2)
        v2 = p1.vectors(p3)
        self.n = v1.tich_co_huong(v2)
        @staticmethod
    def two_planes_corner(m1, m2):
        n1 = m1.n
        n2 = m2.n
        phi = abs(n1.tich_vo_huong(n2)) / (n1.tich_do_dai() * n2.tich_do_dai())
        phi = max(-1, min(1, phi))
        radian_phi = math.acos(phi)
        degree_phi = math.degrees(radian_phi)
        return degree_phi
points = []
for _ in range(4):
    x, y, z = map(float, input().split())
    points.append(Point(x, y, z))
A, B, C, D = points
plane1 = Plane(A, B, C)
plane2 = Plane(B, C, D)
phi = Plane.two_planes_corner(plane1, plane2)
print(f"{phi:.2f}")
```

```python
#W8A12
class ComplexNumber:
    def __init__(self, a, b):
        self.a = a
        self.b = b
    def multiply(self, other):
        real = self.a * other.a - self.b * other.b 
        imaginary = self.a * other.b + self.b * other.a
        return ComplexNumber(real, imaginary)
a1, b1 = map(float, input().split())
a2, b2 = map(float, input().split()) 
z1 = ComplexNumber(a1, b1)
z2 = ComplexNumber(a2, b2)
z3 = z1.multiply(z2)
print(z3.a, z3.b)
```

```python
#W8A13
import math
class Triangle:
    def __init__(self, edge1, edge2, edge3):
        self.a = edge1
        self.b = edge2
        self.c = edge3
    def legit_triangle(self):
        return self.a + self.b > self.c and self.b + self.c > self.a and self.a + self.c > self.b
    def area(self):
        p = (self.a + self.b + self.c) / 2
        return math.sqrt(p * (p - self.a) * (p - self.b) * (p - self.c))
a, b, c = map(float, input().split())
triangle = Triangle(a, b, c)
if not triangle.legit_triangle():
    print("INVALID")
else:
    print(f"{triangle.area():.2f}") 
```
