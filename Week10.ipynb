```python
#W10A3
def HappyNums(n):
    sum = 0
    for i in str(n):
        sum += int(i) ** 2
    return sum 
def check_HappyNums(a):
    check = set()
    while a != 1:
        if a in check:
            return False
        check.add(a)
        a = HappyNums(a)
    return True
try:
    a = int(input())
    print("YES" if check_HappyNums(a) else "NO")
except ValueError:
    print("Value Error")
```

```python
#W10A10
try:
    n = int(input())
    a = list(map(int, input().split()))
    sum = i = 0
    if n == len(a):
        while i + 1 < len(a):
            if a[i] != a[i + 1]:
                sum += a[i]
                i += 1
            else:
                raise ValueError
        sum += a[len(a) - 1]
        print(sum)            
    else:
        raise ValueError
except:
    print("Mang khong hop le")
```

```python
#W10A11
def regex(a):
    reg = {'*', '+', '?'}
    for i in range(len(a)):
        if i == 0 and a[i] in reg:
            return False
        if i > 0 and a[i] in reg and a[i - 1] in reg:
            return False
        if a[i] == '\\':
            if i == len(a) - 1:
                return False
    return True
n = int(input())
for _ in range(n):
    a = input().strip()
    print("True" if regex(a) else "False")
```
