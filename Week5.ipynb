```python
#W5A1
def Max_Of_Two(a, b):
    return max(a, b)
a, b = map(int, input("Enter two numbers: ").split())
print(Max_Of_Two(a, b))
```

```python
#W5A2
def swap(a, b):
    return b, a
a, b = map(int, input("Enter two numbers: ").split())
print(swap(a, b))
```

```python
#W5A3
def Prime_Check(n):
    if n < 2:
        return False
    elif n == 2:
        return True
    else: 
        m = 0
        for i in range(2, int(n**0.5 +2)):
            if n % i == 0 :
                m += 1
        if m != 0:
            return False
        return True
n = int(input("Enter a number: "))
print(Prime_Check(n))
```

```python
#W5A4
def Perfect_Number_Check(n):
    lst = []
    for i in range(1, n):
        if n % i == 0:
            lst.append(i)
    if sum(lst) == n:
        return True
    return False
n = int(input("Enter a number: "))
print(Perfect_Number_Check(n))
```

```python
#W5A5
def List_checking(x, y):
    if y in x:
        return x.index(y)
    return -1
x = list(map(int, input().split()))
y = int(input())
print(List_checking(x, y))
```

```python
#W5A6
def factorial(n):
    m = 1
    for i in range(1, n+1):
        m *= i
    return m
n = int(input("Enter a number: "))
print(factorial(n))
```

```python
#W5A7
def calculator(a, b, n):
    if n == "+":
        return a + b
    elif n == "-":
        return a - b
    elif n == "*":
        return a * b
    elif n == "/":
        return a / b
a, b = map(float, input("Enter two numbers: ").split())
n = input("Enter the operation (+, -, *, /): ")
print(f"result: {calculator(a, b, n)}")
```

```python
#W5A9:
def sum_of_digits(n):
    m = 0
    for i in str(n):
        m += int(i)
    return m
n = int(input("Enter a number: "))
print(sum_of_digits(n))   
```
